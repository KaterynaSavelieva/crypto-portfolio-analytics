```python
import pandas as pd


df = pd.read_csv("data/investments.csv")# Daten aus CSV laden
df
```

```python
from etl.extract import get_bitcoin_price
from etl.transform import transform_data, calculate_profit

price = get_bitcoin_price() # Bitcoin Preis von API holen

df = transform_data(df)     # Daten vorbereiten
df = calculate_profit(df, price)    # Gewinn berechnen

df      # Ergebnis anzeigen
```

```python
df.groupby("client_name")["profit"].sum() # Gewinn pro Kunde berechnen
```

```python
df.groupby("client_name")["profit"].mean()
```

```python
df.groupby("client_name")["profit"].sum().sort_values(ascending=True)
```
